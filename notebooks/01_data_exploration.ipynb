# 01 - Eksploracja danych telemetrii F1

Wczytanie przykładowych danych telemetrycznych (zebranych przez `spark_jobs/streaming_job.py` do `data/processed/telemetry_agg`) i podstawowa eksploracja.

```python
import pandas as pd
import matplotlib.pyplot as plt

DATA_PATH = "/home/jovyan/data/processed/telemetry_agg"
df = pd.read_parquet(DATA_PATH)
df.head()
```

```python
df.describe()
```

```python
for driver_id, g in df.groupby("driver_id"):
    plt.plot(g["distance_m"], g["speed_kmh"], label=driver_id)

plt.xlabel("Distance (m)")
plt.ylabel("Speed (km/h)")
plt.title("Speed trace per driver")
plt.legend()
plt.show()
```
